```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
```

# Gráfica 1

```python
percent_users = [1.45, 1.58, 1.85, 2.00, 2.76, 3.12, 4.05, 4.7, 6.42, 6.72, 6.97, 7.47, 7.83, 8.02, 8.53, 8.85, 9.12]
year = [2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026]

plt.bar(year, percent_users, color=["#00B0F0", "#7030A0"])

# Grafica de histograma
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter())

plt.xlabel("Año")
plt.ylabel("Porcentaje de usuarios")
plt.title("Porcentaje de usuarios por año")

plt.show()
```

#Gráfica 2

```python
year = [2011, 2012, 2013, 2014, 2015, 2016]
bill = [434.52, 345.63, 332.90, 420.50, 457.41, 539.53]

plt.plot(year, bill, color="PINK", marker="o")


plt.gca().set_facecolor('#E7F9FF')  # Color gris claro de fondo

plt.title("PIB Colombiano (mil millones USD) por año")
plt.ylabel("Mil millones de pesos")
plt.xlabel("Años")

plt.show()
```
