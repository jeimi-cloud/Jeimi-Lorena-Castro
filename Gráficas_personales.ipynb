```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
```

# Gráfica 1

```python
percent_users = [1.45, 1.58, 1.85, 2.00, 2.76, 3.12, 4.05, 4.7, 6.42]
year = [2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025, 2026]

plt.bar(year, percent_users, color=["black", "#EAEA43"])

# Esta línea ayuda a formatear los valores del eje Y como porcentajes
plt.gca().yaxis.set_major_formatter(mtick.PercentFormatter())

plt.xlabel("Año")
plt.ylabel("Porcentaje de usuarios")
plt.title("Porcentaje de usuarios por año")

plt.show()
```
